```python
import numpy as np
from datetime import timedelta
import pandas as pd

def generate_temporal_history(row):
    history = []
    base_time = row["timestamp"]
    
    # Define time offsets
    time_offsets = [15, 10, 5, 1]
    
    # Step 1: Create NORMAL baseline (earliest point)
    base_cpu = np.clip(row["cpu_usage"] - np.random.uniform(40, 60), 10, 60)
    base_eff = np.clip(row["energy_efficiency"] + np.random.uniform(0.3, 0.5), 0.6, 1.0)
    base_mem = np.clip(row["memory_usage"] + np.random.uniform(-10, 10), 0, 100)
    
    cpu = base_cpu
    eff = base_eff
    mem = base_mem
    
    # Step 2: Gradually move towards anomaly
    for i, minutes in enumerate(time_offsets):
        temp = row.copy()
        temp["timestamp"] = base_time - timedelta(minutes=minutes)
        
        progress = (i + 1) / len(time_offsets)
        
        # Gradual increase (towards anomaly)
        cpu = cpu + (row["cpu_usage"] - base_cpu) * progress * 0.5 + np.random.uniform(-3, 3)
        eff = eff - (base_eff - row["energy_efficiency"]) * progress * 0.5 + np.random.uniform(-0.05, 0.05)
        mem = mem + np.random.uniform(-5, 5)
        
        temp["cpu_usage"] = np.clip(cpu, 0, 100)
        temp["energy_efficiency"] = np.clip(eff, 0, 1)
        temp["memory_usage"] = np.clip(mem, 0, 100)
        
        history.append(temp)
    
    return pd.DataFrame(history)
```

```python
all_data = []

for _, row in anomaly_dataset.iterrows():
    history = generate_temporal_history(row)
    current = pd.DataFrame([row])
    
    combined = pd.concat([history, current])
    combined = combined.sort_values("timestamp")
    
    all_data.append(combined)

temporal_dataset = pd.concat(all_data, ignore_index=True)
```
