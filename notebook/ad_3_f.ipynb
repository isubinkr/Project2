```python
def get_top_shap_features(sample_df, explainer, top_k=5):
    
    shap_values = explainer.shap_values(sample_df)
    
    if isinstance(shap_values, list):
        shap_values = shap_values[1]
    
    shap_values = shap_values.flatten()
    
    shap_df = pd.DataFrame({
        "feature": sample_df.columns,
        "value": sample_df.values.flatten(),
        "shap_value": shap_values
    })
    
    shap_df["importance"] = shap_df["shap_value"].abs()
    
    top_features = shap_df.sort_values(
        by="importance", ascending=False
    ).head(top_k)
    
    shap_text = ""
    for _, row in top_features.iterrows():
        shap_text += f"{row['feature']} = {row['value']:.2f} (impact={row['shap_value']:.3f})\n"
    
    return shap_text

# CALL SHAP FUNCTION
shap_text = get_top_shap_features(sample_df, explainer)
```

```python
def get_temporal_insights(sample_row, temporal_df):
    
    vm_id = sample_row["vm_id"]
    
    vm_data = temporal_df[temporal_df["vm_id"] == vm_id].copy()
    vm_data = vm_data.sort_values("timestamp")
    
    current_time = vm_data["timestamp"].max()
    
    text = "VM Behavior Timeline:\n\n"
    
    for _, row in vm_data.iterrows():
        label = " (CURRENT STATE)" if row["timestamp"] == current_time else ""
        
        text += f"{row['timestamp']}{label}:\n"
        text += f"  - CPU Usage: {row['cpu_usage']:.2f}\n"
        text += f"  - Memory Usage: {row['memory_usage']:.2f}\n"
        text += f"  - Energy Efficiency: {row['energy_efficiency']:.2f}\n"
        text += f"  - Network Traffic: {row['network_traffic']:.2f}\n"
        text += f"  - Power Consumption: {row['power_consumption']:.2f}\n\n"
    
    return text

# CALL TEMPORAL FUNCTION
temporal_text = get_temporal_insights(sample_row, temporal_df)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
"""
You are a cloud infrastructure monitoring expert.

For VM ID: {vm_id}, the model predicts an anomaly (confidence: {confidence}).

Top contributing features:
{features}

Observed behavior over time:
{temporal_data}

Instructions:
- Be concise and avoid unnecessary explanation.
- Do NOT repeat the same signal (e.g., energy efficiency drop) across sections.
- Focus on relationships between features (e.g., "X increased while Y stayed constant").
- Prioritize insights over description.
- Limit output to the most important points only.

Tasks:
1. Validate if the anomaly prediction is reasonable.
2. Explain the anomaly using key trends and feature interactions.
3. Identify the most likely root cause.
4. Suggest 2-3 specific starting points for investigation.

Output format (strict):

VM ID: <vm_id>

Verdict:
<1-line decision + reason>

Key Observations:
- <concise insight>
- <concise insight>

Root Cause:
<short, specific explanation>

Where to Investigate:
- <action>
- <action>
"""
)

chain = prompt | llm
```

```python
response = chain.invoke({
    "vm_id": sample_row["vm_id"],
    "features": shap_text,
    "temporal_data": temporal_text,
    "confidence": f"{probability:.2f}"
})

print(response.content)
```
